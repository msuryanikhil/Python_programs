#3. check whether the given num is prime or not

```python
#1.
n=int(input())
for i in range(1,n,2):
  print(i, end=" ")
```

```python
#2.
n=int(input())
for i in range(1,11):
  #print("{}*{}={}".format(n,i,n*i))
  print(f"{n}*{i}={n*i}")
```

```python
#3.
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("its not a prime")
    break
else:
  print("yes it is prime")
```

#4.print prime numbers between two given numbers

```python
#4.
n=int(input("n:"))
m=int(input("m:"))
for i in range(n,m+1):
  for j in range(2,i):
     if(i%j==0):
      break
  else:
      print(i)
```

#5.print the sum of the digits of a given number

```python
n=int(input())
sum=0
while(n!=0):
  sum=sum+n%10
  n//=10
print(sum)
```
