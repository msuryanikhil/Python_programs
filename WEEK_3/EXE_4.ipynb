```python
L1=[1,2,3,"hello",4,5,6,7]
L2=[3,4,5.5,6]
print(L1)
print(L1+L2)#concate
print(L1*3)#repetition
print(L1[3])#indexing, print element at index 3
print(3 in L2)#membership
print(L1[2:6])
print(L1[0:])
print(L1[:5])
print(L1[-5:-1])
print(L1[2:8:2])
print(L1[::-1])
for i in L1:
  print(i, end=" ")
```
