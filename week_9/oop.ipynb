## Core Concepts

Object-Oriented Programming (OOP) in Python structures code around **objects** (data + behavior) rather than functions and logic alone.

* **Class:** A blueprint or template for creating objects (e.g., `Car`).
* **Object:** An instance of a class containing real values (e.g., `my_car = Car()`).
* **Attributes:** Variables belonging to a class/object that store state.
* **Methods:** Functions defined inside a class that describe behaviors.

---

## Classes, Objects, and Methods

```python
class Dog:
    # Class attribute (shared by all instances)
    species = "Canis lupus"

    # Constructor method
    def __init__(self, name: str, age: int):
        # Instance attributes (unique to each instance)
        self.name = name
        self.age = age

    # Instance method
    def bark(self) -> str:
        return f"{self.name} says Woof!"


# Creating objects (instantiation)
dog1 = Dog("Buddy", 3)
dog2 = Dog("Milo", 5)

print(dog1.bark())  # Output: Buddy says Woof!
print(dog2.age)   # Output: 5

```

## The Four Pillars of OOP

### 1. Encapsulation

Restricts direct access to an object's data to prevent unintended modification. In Python, access modifiers are indicated by naming conventions:

* **Public:** `self.name` (accessible everywhere)
* **Protected:** `self._balance` (intended for internal/subclass use)
* **Private:** `self.__pin` (name-mangled to prevent direct external access)

```python
class BankAccount:
    def __init__(self, owner: str, balance: float):
        self.owner = owner
        self._account_type = "Checking"  # Protected
        self.__balance = balance          # Private

    def deposit(self, amount: float):
        if amount > 0:
            self.__balance += amount

    def get_balance(self) -> float:
        return self.__balance

account = BankAccount("Alice", 1000)
account.deposit(500)
print(account.get_balance())  # Output: 1500
# print(account.__balance)    # Raises AttributeError

```

### 2. Inheritance

Allows a child class to inherit attributes and methods from a parent class, promoting code reuse.

```python
class Animal:
    def __init__(self, name: str):
        self.name = name

    def speak(self):
        pass

# Child class inheriting from Animal
class Cat(Animal):
    def speak(self) -> str:
        return f"{self.name} says Meow!"

cat = Cat("Whiskers")
print(cat.speak())  # Output: Whiskers says Meow!

```

### 3. Polymorphism

Allows different classes to be treated through a unified interface. Child classes can override methods inherited from parent classes.

```python
class Circle:
    def draw(self):
        return "Drawing a Circle"

class Square:
    def draw(self):
        return "Drawing a Square"

# Polymorphic function handling multiple types
def render_shape(shape):
    print(shape.draw())

render_shape(Circle())  # Output: Drawing a Circle
render_shape(Square())  # Output: Drawing a Square

```

### 4. Abstraction

Hides complex implementation details and enforces specific class structures using abstract base classes (`abc` module).

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass  # Abstract method must be implemented by subclasses

class Car(Vehicle):
    def start_engine(self):
        return "Vroom! Engine started."

# vehicle = Vehicle() # Raises TypeError: Can't instantiate abstract class
my_car = Car()
print(my_car.start_engine())

```

## Class Methods vs Static Methods vs Instance Methods

| Method Type | Decorator | First Argument | Typical Use Case |
| --- | --- | --- | --- |
| **Instance Method** | None | `self` | Accesses/modifies instance state |
| **Class Method** | `@classmethod` | `cls` | Accesses/modifies class-level state or acts as alternative constructor |
| **Static Method** | `@staticmethod` | None | Independent utility functions logically grouped within a class |

```python
class Employee:
    base_salary = 50000

    def __init__(self, name: str):
        self.name = name

    # Instance Method
    def get_details(self):
        return f"Employee: {self.name}"

    # Class Method
    @classmethod
    def set_base_salary(cls, new_salary: int):
        cls.base_salary = new_salary

    # Static Method
    @staticmethod
    def is_work_day(day: str) -> bool:
        return day.lower() not in ["saturday", "sunday"]

```

## Magic Methods (Dunder Methods)

Special methods starting and ending with double underscores (`__`) that let you customize standard Python operations for custom objects:

* `__init__(self, ...)`: Constructor
* `__str__(self)`: Human-readable string representation (used by `print()`)
* `__repr__(self)`: Developer string representation (used in debugging)
* `__len__(self)`: Custom behavior for `len(obj)`
* `__eq__(self, other)`: Custom equality logic (`obj1 == obj2`)

```python
class Book:
    def __init__(self, title: str, pages: int):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"'{self.title}' with {self.pages} pages"

    def __len__(self):
        return self.pages

book = Book("Python 101", 300)
print(str(book)) # Output: 'Python 101' with 300 pages
print(len(book)) # Output: 300

```

## Controlled Attribute Access (`@property`)

Python uses the `@property` decorator to implement getters, setters, and deleters. This allows data validation while keeping attribute access clean (`obj.attr` syntax).

```python
class Temperature:
    def __init__(self, celsius: float):
        self._celsius = celsius  # Protected variable

    @property
    def celsius(self) -> float:
        """Getter for celsius"""
        return self._celsius

    @celsius.setter
    def celsius(self, value: float):
        """Setter with validation"""
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is impossible.")
        self._celsius = value

temp = Temperature(25)
temp.celsius = 30     # Calls setter
print(temp.celsius)   # Calls getter -> Output: 30

```

## Multiple Inheritance & MRO

Python supports multiple inheritance, allowing a class to inherit from multiple parent classes. The **Method Resolution Order (MRO)** determines the search path Python follows to resolve methods.

```python
class A:
    def process(self):
        print("A process")

class B(A):
    def process(self):
        print("B process")
        super().process()

class C(A):
    def process(self):
        print("C process")
        super().process()

class D(B, C):
    def process(self):
        print("D process")
        super().super_process = super().process()

# Check lookup order
print(D.__mro__)
# Output Order: D -> B -> C -> A -> object

```

## Composition vs Inheritance

A fundamental OOP rule is **"Prefer composition over inheritance."**

* **Inheritance ("Is-a" relationship):** A `Dog` *is an* `Animal`.
* **Composition ("Has-a" relationship):** A `Car` *has an* `Engine`.

Composition leads to loosely coupled code that is easier to test and refactor.

```python
class Engine:
    def start(self) -> str:
        return "Engine running..."

class Car:
    def __init__(self, engine: Engine):
        self.engine = engine  # Composition: Car contains an Engine

    def start(self):
        return self.engine.start()

engine = Engine()
my_car = Car(engine)
print(my_car.start())

```

## Modern Python: `dataclasses`

Introduced in Python 3.7, the `dataclass` decorator automatically generates standard boilerplate methods like `__init__()`, `__repr__()`, and `__eq__()`.

```python
from dataclasses import dataclass, field

@dataclass
class Product:
    name: str
    price: float
    quantity: int = 0
    tags: list[str] = field(default_factory=list)

    def total_cost(self) -> float:
        return self.price * self.quantity

item = Product(name="Laptop", price=999.99, quantity=2)
print(item)  # Product(name='Laptop', price=999.99, quantity=2, tags=[])
print(item.total_cost())  # Output: 1999.98

```

## Core Pillars of Object Orientation

### 1. Encapsulation

Bundling data and the methods that manipulate that data inside a single unit (a class), while restricting direct access to internal details.

* **Goal:** Protects state integrity and prevents unauthorized external modifications.

### 2. Abstraction

Hiding complex implementation details and presenting only essential features to the outside world.

* **Goal:** Reduces complexity by allowing users to interact with high-level interfaces without needing to know how operations work under the hood.

### 3. Inheritance

Allowing a new class (child/subclass) to acquire attributes and behaviors from an existing class (parent/superclass).

* **Goal:** Promotes code reuse and establishes logical hierarchies.

### 4. Polymorphism

The ability for different classes to respond to the same method or interface call in their own unique way.

* **Goal:** Enables flexibility—you can write generic code that works across multiple distinct types.

## Key Terminology

* **Class:** The structural blueprint or template defining default properties and behaviors.
* **Object / Instance:** A concrete realization of a class allocated in memory with specific data values.
* **Attributes / Fields:** Variables bound to an object or class that store its state.
* **Methods:** Functions defined within a class that operate on an object's state or perform tasks related to it.
* **Constructor:** A specialized method executed automatically when an object is created to initialize its state.

---

## Design Concepts & Principles

* **Composition over Inheritance:** A structural design principle suggesting classes should achieve complex behavior by referencing other objects (*"has-a"*) rather than inheriting from base classes (*"is-a"*).
* **SOLID Principles:** Five fundamental object-oriented design rules (Single Responsibility, Open/Closed, Liskov Substitution, Interface Segregation, Dependency Inversion) aimed at making software readable, maintainable, and scalable.
