#Object Orientation Programming through python

```python
#class creation
class Student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def display(self):
    print(self.name, self.marks)
#object creation
s1= Student("sai",90)
s2= Student("ravi",89)
s2.display()
s1.display()
```
