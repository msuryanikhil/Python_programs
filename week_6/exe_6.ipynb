#palindrome number checking

```python
def palindrome(n):
  temp=n
  rev=0
  while n>0:
    rev=rev*10+n%10
    n=n//10
  if temp==rev:
    print("palindrome")
  else:
    print("not palindrome")
palindrome(int(input()))
```

#program that reverse the string using functions

```python
def reverse(s):
  return s[::-1]
print(reverse(input()))
```

```python
#without indexing
def reverse(s):
  str=""
  for i in s:
    str=i+str
  return str
print(reverse(input()))
```

```python
#without indexing
def reverse(s):
  str=""
  for i in range(len(s)-1,-1,-1):
    str=str+s[i]
  return str
print(reverse(input()))
```

#types of arguments

```python
#default arguments
def printMSG(name, branch ="AI&DS"):
  print("name is", name, "and branch is", branch)
printMSG("sai")
```

```python
#variable length arguments
def add(*n):
  print(sum(n))
  print(type(n))
add(1,2,3,4,5)
```

```python
#function that returns multiple
def multiple(*n):
  s=sum(n)
  c=len(n)
  avg=s/c
  return s,c,avg
x,y,z=multiple(*map(int, input().split()))
print(f"sum is {x}, count is {y}, average is {z}")
```
