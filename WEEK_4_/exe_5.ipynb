#program that count the no of vowels and consonants in a string

```python
a=input()
vowels=0
consonants=0
s=a.lower()
for i in s:
  if i in "aeiou":
    vowels+=1
  elif i.isalpha():
    consonants+=1
print(vowels)
print(consonants)
```
