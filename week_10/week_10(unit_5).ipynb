```python
import pandas as pd

# Let's create a sample DataFrame with 'attendence' and 'marks' columns
data = {
    'attendence': [60, 75, 80, 85, 90, 95, 100],
    'marks': [50, 65, 70, 78, 85, 92, 98]
}
df = pd.DataFrame(data)
display(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

attendence = df["attendence"]
marks = df["marks"]

# Scatter plot
plt.figure()
plt.scatter(attendence, marks)
plt.xlabel("attendence")
plt.ylabel("marks")
plt.title("scatter plot")
plt.show()
```
